# Bike Rental Data Analysis

This notebook demonstrates a basic exploratory data analysis (EDA) of the bike rental dataset.
It loads the cleaned CSV, shows a preview, and visualises key trends such as rentals over time and popular stations.

```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

# Load the cleaned bike rental data
df = pd.read_csv('bike_rental_cleaned.csv')
df.head()
```

```python
# Basic info
print('Rows:', len(df))
print('Columns:', len(df.columns))
df.describe(include='all')
```

```python
# Plot rentals over time (assuming a 'date' column)
if 'date' in df.columns:
    df['date'] = pd.to_datetime(df['date'])
    df.set_index('date', inplace=True)
    daily = df.resample('D').size()
    daily.plot(title='Daily Bike Rentals')
    plt.ylabel('Number of rentals')
    plt.show()
else:
    print('No date column found for time series plot.')
```
